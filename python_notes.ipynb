# Introduction to Problem Solving

Problem solving in programming means converting a real-world problem into a step-by-step logical solution.

## Steps for Problem-Solving

### 1. Analyzing the Problem

* Understand **what is given (input)** and **what is required (output)**
* Identify constraints and conditions

* *Example:*
    Find sum of two numbers
    * Input  → A, B
    * Output → A + B
---

### 2. Developing an Algorithm

* A **step-by-step logical procedure/solution** to solve the problem
* Written in simple language (not code)

    *Example Algorithm (Sum):*
    1. Start
    2. Read A, B
    3. Sum = A + B
    4. Display Sum
    5. End
---

### 3. Coding

* Convert algorithm into programming language (like Python, C, Java)

    **Example (Python):**
    ```
    A = int(input("Enter the value of A: "))
    B = int(input("Enter the value of B: "))
    print("Sum = ", A + B)
    ```

### 4. Testing

* Check program with different inputs
* Ensure correct output

    **Example Test Cases:**
    * Input: 2, 3 → Output: 5
    * Input: -1, 5 → Output: 4
---

### 5. Debugging

* Finding and fixing errors (bugs)

    **Types of Errors:**

    * Syntax Error (wrong code format)
    * Logical Error (wrong logic)
    * Runtime Error (crashes during execution)

# Chapter Exception Handling

## **1. Introduction to Errors**
In Python, programs may fail to execute correctly due to three types of errors:
* **syntax errors**
* **runtime errors**
* **logical errors**.

*   **Syntax Errors (Parsing Errors):** These occur when the programmer does not follow the rules of the Python language. The interpreter detects these before execution and will not run the program until they are fixed.

*   **Exceptions:** These are errors that occur during the execution of a syntactically correct statement, such as trying to divide by zero or opening a file that does not exist. An exception is a **Python object** that represents an error.

## **2. Built-in Exceptions**
Python’s standard library provides standardized solutions for common errors. Some frequently encountered built-in exceptions include:

| Exception Name | When it is raised |
| :--- | :--- |
| **`SyntaxError`** | Error in the syntax of the code. |
| **`ValueError`** | A function receives an argument of the correct type but an inappropriate value. |
| **`ZeroDivisionError`** | The denominator in a division operation is zero. |
| **`IndexError`** | A sequence subscript (index) is out of range. |
| **`NameError`** | A local or global variable name is not defined. |
| **`TypeError`** | An operator is supplied with a value of the incorrect data type. |
| **`IOError`** | A specified file cannot be opened. |
| **`IndentationError`** | Incorrect indentation in the code. |
| **`EOFError`** | reached at end of file |

## **3. Raising Exceptions Forcefully**
Programmers can manually trigger exceptions using two specific statements:

*   **The `raise` Statement:** Used to "throw" an exception manually.
    *   *Syntax:* `raise exception-name[(optional string argument)]`.
    *   *Example:* `raise IndexError("OOPS : An Exception has occurred")`.
*   **The `assert` Statement:** Used to test an expression. If the expression is **false**, an `AssertionError` is raised. It is often used at the beginning of functions to check for valid input.
    *   *Syntax:* `assert Expression[, arguments]`.
    *   *Example:* `assert(number >= 0), "OOPS... Negative Number"`.


## **4. The Exception Handling Process**
When an error occurs, the interpreter creates an **exception object** containing error details (type, file name, and position).
1.  **Throwing:** The object is handed to the runtime system.
2.  **Searching:** The system searches the **call stack** (the list of methods called) in reverse order for a handler.
3.  **Catching:** Executing the appropriate handler code is called **catching the exception**. If no handler is found, the program terminates.


## **5. Handling Blocks: `try`, `except`, `else`, and `finally`**

### **The `try...except` Block**
Suspicious lines of code are placed in the `try` block, and the code to handle errors is placed in the `except` block.
*   **Example (Single Exception):**
    ```python
    try:
        denom = int(input("Enter denominator: "))
        print(50 / denom)
    except ZeroDivisionError:
        print("Denominator as ZERO.... not allowed")
    ```

## **Multiple `except` Clauses**
A single `try` block can have multiple `except` blocks to handle different types of errors. You can also use a **generic `except` block** (without a name) as a catch-all for any unnamed exceptions.
*   **Example:**
    ```python
    try:
        # code here
    except ZeroDivisionError:
        print("Zero error")
    except ValueError:
        print("Only integers allowed")
    except:
        print("Some other exception raised")
    ```

## **The `else` Clause**
This optional block executes **only if no exception** occurred in the `try` block.
*   **Example:**
    ```python
    try:
        quotient = 50 / denom
    except ZeroDivisionError:
        print("Error")
    else:
        print("Result is", quotient)
    ```


## **The `finally` Clause**
This block **always executes**, regardless of whether an exception occurred or was handled. It is commonly used for cleanup tasks like closing files.
*   **Example:**
    ```python
    try:
        # code
    finally:
        print("OVER AND OUT")
    ```
*   **Note:** Unlike `except`, the `finally` block does not terminate the exception. If an exception is not handled by any `except` block, it is **re-raised** after `finally` finishes executing.

# **Chapter : Functions in Pythyon**

## **1. Why do we need Functions? (The "Modularity" Concept)**
As programs grow complex, the number of lines increases, making them bulky and difficult to manage.
*   **Modularity:** This is the process of dividing a large program into separate, independent blocks or sub-problems, each with a specific name and functionality.
*   **Reusability:** Once you write a function, you can call it multiple times from anywhere in your program without rewriting the code.
*   **Other Benefits:** Functions increase readability, reduce code length, make debugging easier, and allow teams to work on different parts of a program in parallel.

## **2. Types of Functions**
According to the syllabus, there are three main types:
1.  **Built-in Functions:** Ready-made functions already defined in the Python interpreter (e.g., `input()`, `int()`, `print()`, `len()`).
2.  **Functions Defined in Modules:** Functions that belong to specific files (modules) and need to be imported before use (e.g., `math.sqrt()`, `random.randint()`).
3.  **User-Defined Functions:** Functions created by the programmer to perform specific tasks as per their requirements.

## **3. Creating a User-Defined Function**
To create your own function, you use the `def` (short for define) keyword.

**The Syntax:**
```python
def function_name([parameters]):
    # Function Body (indented statements)
    return [value]
```

**Key Rules for Creation:**
*   **The Header:** Must start with `def` and **always end with a colon (:)**.
*   **Naming:** The name must be unique and follow the same rules as identifiers (variables).
*   **Indentation:** Statements inside the function must be indented. Anything outside the indentation is not part of the function.
*   **Parameters:** These are optional variables inside the brackets that receive values when the function is called.

## **4. Arguments and Parameters**
These two terms are often confused, but here is the simple difference:
*   **Parameter:** The variable defined in the **function header** (e.g., `n` in `def sumSquares(n):`).
*   **Argument:** The actual **value passed** to the function during the call (e.g., `5` in `sumSquares(5)`).

**Cool Fact:** Both the argument and the parameter refer to the same value in memory and will have the same `id()` until the value is modified.


## **5. Types of Parameters**

### **A. Positional Parameters**
Arguments must be provided in the **same order** as the parameters in the function header. If you swap the order of arguments, the function may produce incorrect results or errors.

### **B. Default Parameters**
Python allows you to assign a "predecided" value to a parameter in case the user forgets to provide an argument during the call.
*   **Example:** `def mixedFraction(num, deno = 1):`.
*   **The "Trailing" Rule:** If a parameter has a default value, all parameters to its **right** must also have default values.
    *   *Correct:* `def calc(rate, time=5, principal=1000):`.
    *   *Incorrect:* `def calc(principal=1000, rate, time=5):` (because `rate` has no default but is to the right of `principal`).


## **6. Function Returning Value(s)**
A function can send data back to the place where it was called using the `return` statement.
*   **Void Functions:** Functions that perform a task but do not return any value (they actually return `None`).
*   **Returning Multiple Values:** This is a special Python feature! You can return multiple values separated by commas. Python automatically packages them into a **Tuple**.
    *   *Example:* `return (area, perimeter)`.


## **7. Flow of Execution**
This is the order in which statements are executed.
1.  Execution starts from the first statement and goes top-to-bottom.
2.  **Function definitions** are noted by the interpreter but **not executed** until they are called.
3.  When a **function call** is reached, the control **jumps** to the function definition, runs the code inside, and then **jumps back** to where it left off.
4.  **CRITICAL RULE:** A function must be defined **before** it is called in the program, otherwise, you get a `NameError`.


## **8. Scope of a Variable**
Scope refers to the part of the program where a variable is "visible" or accessible.

*   **Local Scope:** A variable defined **inside** a function. It exists only while that function is running and cannot be accessed outside it.
*   **Global Scope:** A variable defined **outside** all functions. It is accessible throughout the entire program.
*   **The `global` Keyword:** If you want to modify a global variable from inside a function, you must prefix the variable name with the `global` keyword. Without it, Python will just create a new local variable with the same name.

## **9. Python Standard Library (Modules)**
A **Module** is a grouping of related functions saved in a `.py` file.

*   **`import modulename`:** Gives you access to all functions. You must use the dot (.) separator (e.g., `math.sqrt()`).
*   **`from modulename import functionname`:** Imports only specific functions. You can call them directly without the module name (e.g., just `sqrt()`).
*   **Common Modules:**
    *   **`math`:** `ceil()`, `floor()`, `fabs()`, `factorial()`, `pow()`, `sqrt()`, `gcd()`, `sin()`.
    *   **`random`:** `random()` (0.0 to 1.0), `randint(x, y)` (inclusive integer), `randrange()`.
    *   **`statistics`:** `mean()`, `median()`, `mode()`.

## **Quick Summary for Boards:**
*   **Function:** Named group of instructions for a specific task.
*   **Default values** must be at the end of the parameter list.
*   **Multiple values** are returned as a **Tuple**.
*   **Functions** must be defined before calling.
*   **`global`** keyword is needed to change global variables inside a function.

# **Chapter: File Handling in Python**

## **1. Introduction to Files**
In our previous programs, data was stored in variables. However, variables only live while the program is running. To store data **permanently** so it can be reused later, we use **Files**.
*   **Definition:** A file is a named location on secondary storage media (like a hard disk) where data is stored permanently.
*   **The File Object (Handle):** To work with a file in Python, we create a "File Object." This acts as a link or a bridge between your program and the data file stored on the disk.


## **2. Types of Files**
Computers store everything as 0s and 1s, but we categorize them into two main types:

| Feature | **Text File** | **Binary File** |
| :--- | :--- | :--- |
| **Readability** | Human-readable characters (alphabets, numbers). | Non-human readable (images, audio, video). |
| **Storage** | Stored as ASCII or Unicode values. | Stored as a stream of raw bytes. |
| **EOL** | Each line ends with an **End of Line (EOL)** character (default is `\n`). | No specific EOL character. |
| **Extension** | `.txt`, `.py`, `.csv`. | `.dat`, `.exe`, `.jpg`, `.mp3`. |


## **3. Working with Text Files**



### **A. Opening and Closing**
To open a file, use the `open()` function.
*   **Syntax:** `file_object = open("filename", "access_mode")`.
*   **Modes:**
    *   `'r'`: Read-only (Default). Starts at the beginning.
    *   `'w'`: Write-only. **Overwrites** existing data. Creates a new file if it doesn't exist.
    *   `'a'`: Append. Adds new data to the **end** of the file.
    *   `'r+'`, `'w+'`, `'a+'`: Used for both reading and writing.

**Closing a file:** Use `file_object.close()`. This frees up memory and ensures all data in the "buffer" is actually saved to the disk.

### **B. The `with` Clause (The Smart Way)**
Opening a file using `with` is better because it **automatically closes** the file when you are done, even if an error occurs.
*   **Example:**
    ```python
    with open("notes.txt", "r") as f:
        data = f.read()
    # File closes automatically here!
    ```

### **C. Writing Data**
1.  **`write(string)`:** Writes a single string to the file. You must add `\n` manually for new lines.
2.  **`writelines(list)`:** Writes a list of strings to the file.

### **D. Reading Data**
1.  **`read(n)`:** Reads `n` bytes. If `n` is missing, it reads the **entire file**.
2.  **`readline()`:** Reads one complete line at a time.
3.  **`readlines()`:** Reads all lines and returns them as a **list of strings**.

## **4. Setting Offsets (Moving the Pointer)**
Python keeps track of where you are in a file using a "pointer" or "offset".
*   **`tell()`:** Tells you the current byte position of the file pointer.
*   **`seek(offset, reference_point)`:** Moves the pointer to a specific location.
    *   `reference_point = 0`: From the **beginning** of the file (Default).
    *   `reference_point = 1`: From the **current** position.
    *   `reference_point = 2`: From the **end** of the file.


## **5. Binary Files and the `pickle` Module**
Binary files store actual Python objects (like lists or dictionaries). To do this, we use **Serializing** (Pickling) and **Deserializing** (Unpickling).

*   **Pickling (`dump`):** Converting a Python object into a byte stream to save it.
    *   **Syntax:** `pickle.dump(object, file_object)`.
*   **Unpickling (`load`):** Converting a byte stream back into a Python object.
    *   **Syntax:** `object = pickle.load(file_object)`.

**Important:** Always use binary modes like `'wb'` (write binary) or `'rb'` (read binary) when working with the `pickle` module.


## **6. CSV Files (Comma Separated Values)**
CSV files are text files where data is separated by commas, making them easy to open in Excel.
*   **Module:** `import csv`.
*   **Writing:**
    *   `writer()`: Creates a writer object.
    *   `writerow(list)`: Writes a single row.
    *   `writerows(nested_list)`: Writes multiple rows at once.
*   **Reading:**
    *   `reader()`: Used to read data row by row from the file.

### **Quick Revision Summary**
*   **Text files** are human-readable; **Binary files** are byte-streams.
*   **`open()`** returns a handle; **`close()`** saves changes and clears memory.
*   **`'w'`** kills old data; **`'a'`** adds to the end.
*   **`tell()`** is "Where am I?"; **`seek()`** is "Take me there".
*   **Pickle `dump()`** is for saving; **`load()`** is for reading binary objects.
*   **CSV `writerow()`** writes a list as a row in a spreadsheet-like file.
